```python
# 딕셔너리 표현식 사용하기 (딕셔너리 컴프리헨션)

## 1. {키 : 값 for 키, 값 in 딕셔너리}
## 2. dict({키 : 값 for 키, 값 in 딕셔너리})

keys = ["a", "b", "c", "d"]
x = {key : value for key, value in dict.fromkeys(keys).items()}
print(x)
```

```python
keys = ["a", "b", "c", "d"]
{key : 0 for key in dict.fromkeys(keys).keys()}
```

```python
# 딕셔너리 표현식에서 if 조건문 사용하기
x = {
    'a' : 10,
    'b' : 20,
    'c' : 30,
    'd' : 40
}

# for key, value in x.items():
#     if value == 20: <- 반복문이 돌아가는 중에 원본 x가 지워지기 때문에 에러
#         del x[key]

for key, value in list(x.items()): # list(x.items())는 원본 외에 다른 리스트
    if value == 20: # list(x.items())에서 지워지는거라 정상 작동
        del x[key]     
        
print(x)
```

```python
x = {
    'a' : 10,
    'b' : 20,
    'c' : 30,
    'd' : 40
}

{key: value for key, value in x.items() if value != 20}
```

```python
# 딕셔너리의 할당과 복사
x = {'a' : 0, 'b' : 0, 'c' : 0, 'd' : 0}
y = x
```

```python
print(x is y)
print(id(x))
print(id(y))
```

```python
y["a"] = 99
print(x)
print(y)
```

```python
x = {'a' : 0, 'b' : 0, 'c' : 0, 'd' : 0}
y = x.copy()

y['a'] = 99

print(x is y)
print(x)
print(y)
```

```python
x = {'a' : {'python': '2.7'}, 'b' : {'python' : '3.6'}}
y = x.copy()

y['a']['python'] = '2.7.15'

print(x)
print(y)
```

```python
import copy

x = {'a' : {'python': '2.7'}, 'b' : {'python' : '3.6'}}
y = copy.deepcopy(x)

y['a']['python'] = '2.7.15'

print(x)
print(y)
```
