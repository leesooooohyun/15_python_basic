```python
# 위치 인수를 사용하는 함수를 만들고 호출하기
def print_numbers(a, b, c):
    print(a)
    print(b)
    print(c)

print_numbers(10, 20, 30)
```

```python
# 언패킹 사용하기
print_numbers(*[10, 20, 30])
```

```python
# 가변 인수 만들기
## def 함수이름(*매개변수):
##      코드

def print_number(*args):
    print(args)
    for arg in args:
        print(arg)
```

```python
print_number(10)
print_number(10, 20, 30, 40)
```

```python
x = [10]
print_number(*x)

y = [10, 20, 30, 40]
print_number(*y)
```

```python
def print_numbers(a, *args):
    print(a)
    print(args)

print_numbers(1)
print_numbers(1, 10, 20)
print_numbers(*[10, 20, 30])
```

```python
# 키워드 인수 사용하기

def personal_info(name, age, address):
    print('이름: ', name)
    print('나이: ', age)
    print('주소: ', address)

personal_info('홍길동', 30, '서울시 용산구')
```

```python
personal_info(name = '홍길동', age = 30, address = '서울시 용산구')
```

```python
personal_info(age = 30, address = '서울시 용산구', name = '홍길동')
```

```python
x = {
    'name' : '홍길동',
    'age' : 30,
    'address' : '서울시 용산구'
}

personal_info(**x) # 딕셔너리는 키-값 형태이기 때문에 * 2개 쓰면 언패킹 가능
```

```python
# 키워드 인수를 사용하는 가변 인수 함수 만들기

## def 함수이름(*매개변수):
##      코드

def personal_info(**kwargs):
    print(kwargs)
    for kw, arg in kwargs.items():
        print(kw, ': ', arg, sep='')
```

```python
personal_info(name = '홍길동')
```

```python
personal_info(age = 30, address = '서울시 용산구', name = '홍길동')
```

```python
# 연습문제 30.6

korean, english, mathmatics, science = 100, 86, 81, 91

def get_max_score(*args):
    return max(args)

max_score = get_max_score(korean, english, mathmatics, science)
print("높은 점수: ", max_score)

max_score = get_max_score(english, science)
print("높은 점수: ", max_score)
```

```python
# 심사문제 30.7

korean, english, mathmatics, science = map(int, input().split())

def get_min_max_score(*args):
    return min(args), max(args)

def get_average(**kargs):
    scores = kargs.values()
    return sum(scores) / len(scores)

min_score, max_score = get_min_max_score(korean, english, mathmatics, science)
average_score = get_average(korean=korean, english=english, mathmatics=mathmatics, science=science)
print('낮은 점수: {0: .2f}, 높은 점수: {1: .2f}, 평균 점수: {2: .2f}'.format(min_score, max_score, average_score))

min_score, max_score = get_min_max_score(english, science)
average_score = get_average(english=english, science=science)
print('낮은 점수: {0: .2f}, 높은 점수: {1: .2f}, 평균 점수: {2: .2f}'.format(min_score, max_score, average_score))
```

```python
prices = {
    "아메리카노": 2500,
    "카페라떼": 3500,
    "바닐라라떼": 4000,
    "초코케이크": 5500,
    "샌드위치": 6000
}


def order(**kwargs):
  total_price = 0
  for key, value in kwargs.items():
    # menu_price = 0
    if key in prices:
      menu_price = prices[key] * value
      print(f"{key} {value}개: {menu_price}원")
      total_price += menu_price
    else:
      print(f"등록되지 않은 상품입니다: {key}")
  print(f"총 결제 금액: {total_price}원")
  print("-" * 20)

order(아메리카노=2, 카페라떼=1)
order(바닐라라떼=2, 초코케이크=1, 샌드위치=1)
order(아메리카노=1, 허니브레드=2)
```
