```python
def hello(count):
    if count == 0:
        return
    
    print("hello world", count)
    count -= 1
    hello(count)
    
```

```python
hello(5)
```

```python
# 재귀호출로 팩토리얼 구하기

def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)
```

```python
factorial(5)
```

```python
# 연습문제 31.4

def is_palindrome(word):
    if len(word) < 2: # 한 글자면 True
        return True
    if word[0] != word[-1]:
        return False
    return is_palindrome(word[1:-1])


print(is_palindrome('hello'))
print(is_palindrome('level'))
```

```python
# 심사문제 31.5

def fin(n):
    # 종료 조건: n이 0이거나 1일 때는 그 값(n)을 그대로 반환하여 재귀를 멈춤
    if n <= 1:
        return n
    
    # 재귀 호출: 피보나치 공식(앞의 두 수의 합)을 그대로 반환
    return fin(n - 1) + fin(n - 2)

n = int(input())
print(fin(n))
```
