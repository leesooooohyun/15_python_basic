```python
# 집합연산 사용하기
## 세트1 | 세트2
# set.union(세트1, 세트2)


a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

#합집합

print(a | b)
print(set.union(a, b))
print(a.union(b))
```

```python
# 교집합
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a & b)
print(set.intersection(a, b))
print(a.intersection(b))
```

```python
# 차집합
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a - b)
print(set.difference(a, b))
print(a.difference(b))
```

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(b - a)
print(set.difference(b, a))
print(b.difference(a))
```

```python
# 대칭차집합
## 세트1 ^ 세트2
## set.symmetric_difference(세트1, 세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a ^ b)
print(set.symmetric_difference(a, b))
print(a.symmetric_difference(b))
```

```python
# 집합 연산 후 할당 연산자 사용하기

## 세트1 |= 세트2
## 세트1.update(세트2)

a = {1, 2, 3, 4}
a |= {5}

print(a)

a.update({6})
print(a)
```

```python
a = {1, 2, 3, 4}
a.intersection({3, 4}) # 이대로는 재할당이 안됨. a가 안 바뀜 
a.intersection_update({3, 4}) # 재할당
print(a)
```

```python
## 세트1 &= 세트2
## 세트1.intersection_update(세트2)

a = {1, 2, 3, 4}
a &= {0, 1, 2, 3, 4}
print(a)

a = {1, 2, 3, 4}
a.intersection_update({0, 1, 2, 3, 4})
print(a)
```

```python
## 세트1 ^= 세트2
## 세트1.symmetric_difference_update(세트2)

a = {1, 2, 3, 4}
a ^= {3, 4, 5, 6}
print(a)

a = {1, 2, 3, 4}
a.symmetric_difference_update({3, 4, 5, 6})
print(a)
```

```python
# 부분집합과 상위집합 확인하기
## 현재세트 <= 다른세트
## 현재세트.issubset(다른세트)

a = {1, 2, 3, 4}
print(a <= {1, 2, 3, 4})

print(a.issubset({1, 2, 3, 4, 5}))
```

```python
## 현재세트 < 다른세트
a = {1, 2, 3, 4}
a < {1, 2, 3, 4, 5}
```

```python
## 현재세트 >= 다른세트
## 현재세트.issuperset(다른세트)

a = {1, 2, 3, 4}
print(a >= {1, 2, 3, 4})

print(a.issuperset({1, 2, 3, 4}))
```

```python
## 현재세트 > 다른세트

a = {1, 2, 3, 4}
a > {1, 2, 3}
```

```python
# 세트가 같은지 다른지 확인하기
a = {1, 2, 3, 4}
print(a == {1, 2, 3, 4})

print(a == {4, 2, 1, 3}) # 순서 상관없음
```

```python
# 세트가 겹치지 않는지 확인하기
## 현재세트.isdisjoint(다른세트)
## 겹치는 요소가 있으면 False, 없으면 True

a = {1, 2, 3, 4}
print(a.isdisjoint({5, 6, 7, 8})) # 겹치는 요소 없음
print(a.isdisjoint({3, 4, 5, 6})) # 3, 4 겹침
```

```python
# 세트 표현식에 if 조건문 사용하기
## 1. {식 for 변수 in 세트 if 조건식}
## 2. set(식 for 변수 in 세트 if 조건식)

a = {i for i in "pineapple" if i not in "apl"}
print(a)
```

```python
# 심사문제 26.9
first, second = map(int, input().split())

a = {i for i in range(1, first + 1) if first % i == 0}
b = {i for i in range(1, second + 1) if second % i == 0}

divisor = a & b # a, b 교집합

result = 0 # 교집합 다 더하기
if type(divisor) == set:
    result = sum(divisor)

print(result)
```
