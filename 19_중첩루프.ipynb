```python
for i in range(5):
    for j in range(5):
        print("j: ", j, sep='', end= ' ')
    print("i: ", i, "\\n", sep="")
```

```python
# 사각형으로 별 출력하기

for i in range(5):
    for j in range(5):
        print("*", end='')
    print()
```

```python
# 사각형 모양 바꾸기
for i in range(3):
    for i in range(7):
        print("*", end='')
    print()
```

```python
# 계단식으로 열 출력하기

for i in range(5):
    for j in range(5):
        if j < i:
            print("*", end='')
    print()
```

```python
# 대각선으로 별 출력하기
for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end='')
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end='')
        else:
            print(" ", end="")
    print()
```

```python
# 연습문제 19.5
for i in range(5):
    for j in range(5):
        if j >= i:
            print("*", end='')
        else:
            print(" ", end="")
    print()
```

```python
# 심사문제 19.6

height = int(input())

# for i in range(height):
for i in range(1, height + 1):
    print(" " * (height - i), end='')
    print("*" * (2 * i - 1))
```

```python
height = int(input())

for i in range(1, height + 1):
    for j in range(height - i): # 공백 출력
        print(" ", end='')
    for j in range(2 * i - 1): # 별 출력
        print("*", end="")
    print()
    
```
