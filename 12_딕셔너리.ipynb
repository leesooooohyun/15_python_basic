```python
# 딕셔너리 만들기

lux = {
    "health" : 490,
    "mana" : 334,
    "melee" : 550,
    "armor" : 18.72
}

print(type(lux))
print(lux)
```

```python
# 키 이름이 중복이 되면?
lux = {
    "health" : 490,
    "health" : 800,
    "mana" : 334,
    "melee" : 550,
    "armor" : 18.72
}
print(lux)
print(lux['health'])
```

```python
# 딕셔너리 키의 자료형
## 1. 문자형 가능
## 2. 정수형 가능
## 3. 실수형 가능
## 4. 불(True, False) 가능
## 5. 튜플형 가능

## 불가능한 키의 자료형
## 1. 딕셔너리 불가능
## 2. 리스트 불가능
```

```python
x = {
    100 : "hundred",
    False : 0,
    3.5 : [3.5, 3.5]
}

print(type(x))
print(x)
```

```python
# 리스트를 딕셔너리 키로 사용 할 수 없음
y = {
    [10, 20] : 100
}
print(y)
print(type(y))
```

```python
# 딕션리를 딕셔너리 키로 사용 할 수 없음
z = {
    {"a" : 10} : 100
}
print(z)
print(type(z))
```

```python
# 빈 딕셔너리 만들기
x = {}
y = dict()

print(type(x))
print(type(y))
print(x)
print(y)
print(bool(x)) # 비어있기 때문에 0 -> False
print(bool(y))
```

```python
# dict로 딕셔너리 만들기
lux1 = dict(health=400, mane=334, melee=550, armor=10.72)
print(lux1)
print(type(lux1))
```

```python
lux2 = dict(zip(["health", "mane", "melee", "armor"], [490, 334, 550, 10.72]))
print(lux2)
print(type(lux2))
```

```python
lux3 = dict([("health", 490), ("mane", 334), ("melee", 550), ("armor", 10.72)])
```

```python
# 딕셔너리의 키에 값 할당하기
lux = {
    "health" : 490,
    "mana" : 334,
    "melee" : 550,
    "armor" : 18.72
}

print(lux['health'])
print(lux['armor'])
```

```python
lux['health'] = 2037
print(lux)
```

```python
lux["mana_regen"] = 3.28
print(lux)
```

```python
# 딕셔너리에 키가 있는지 확인하기
print(lux)

"health" in lux
"attack_speed" in lux
```

```python
# 딕셔너리 키의 개수 구하기
print(len(lux))
```
