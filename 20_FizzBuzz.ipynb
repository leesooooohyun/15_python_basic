```python
# 1부터 100까지 출력하기
for i in range(100):
    print(i)
```

```python
# 3의 배수 Fizz, 5의 배수 Buzz
for i in range(100):
    if i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 3과 5의 공배수 처리하기
for i in range(100):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 논리연산자를 사용하지 않고 3과 5의 공배수 처리하기
for i in range(100):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 코드 단축하기
for i in range(1, 101):
    print('Fizz' * (i % 3 == 0) + 'Buzz' * (i % 5 == 0) or i)
```

```python
# 연습문제 20.7

for i in range(1, 101):
    if i % 22 == 0:
         print("FizzBuzz")
    elif i % 2 == 0:
        print("Fizz")
    elif i % 11 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 심사문제 20.8

first, second = map(int, input().split())

if first < second and 1 <= first <= 1000 and 10 <= second <= 1000:
    for i in range(first, second + 1):
        if i % 35 == 0:
            print("FizzBuzz")
        elif i % 5 == 0:
            print("Fizz")
        elif i % 7 == 0:
            print("Buzz")
        else:
            print(i)    
else:
    print("잘못된 입력입니다.")
```
