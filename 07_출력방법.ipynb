```python
# 연습문제 7.4

year = 2000
month = 10
day = 27
hour = 11
minute = 43
second = 59

print(year, month, day, sep="/", end=' ')
print(hour, minute, second, sep=':')
```

```python
# 심사문제 7.5
year, month, day, hour, minute, second = input().split()

print(year, month, day, sep='-', end='T') # .zill(2) 쓰면 2자리수에 맞춰 0을 채워줌
print(hour, minute, second, sep=':')
```
