```python
import turtle as t

t.shape("turtle")
for i in range(4):
    t.forward(100)
    t.right(90)
```

```python
# 오각형 만들기

for i in range(5):
    t.forward(100)
    t.right(360 / 5)
```

```python
# 육각형 만들기

for i in range(6):
    t.forward(100)
    t.right(360 / 6)
```

```python
# 다각형에 색칠하기
import turtle as t

n = 6
t.shape("turtle")
t.color("red")
t.begin_fill()
for i in range(n):
    t.forward(100)
    t.right(360 / n)
t.end_fill()
```

```python
# 원을 반복해서 그리기
n = 60

t.speed("fastest")
for i in range(n):
    t.circle(120)
    t.right(360/n)
```
