```python
# 클래스로 점 구현하기

class Point2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y


p1 = Point2D(x=30, y=20)
p2 = Point2D(x=60, y=50)

print(f"p1 : {p1.x} {p1.y}")
print(f"p2 : {p2.x} {p2.y}")
```

```python
import math

class Point2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y


p1 = Point2D(x=30, y=20)
p2 = Point2D(x=60, y=50)

a = p2.x - p1.x
b = p2.y - p1.y

# c = math.sqrt((a ** 2) + (b ** 2))
c = math.sqrt(pow(a, 2) + pow(b, 2))
print(c)
```
