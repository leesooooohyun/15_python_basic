```python
# 논리 연산자 사용하기
#AND
print(True and True)
print(True and False)
print(False and True)
print(False and False)
```

```python
#OR
print(True or True)
print(True or False)
print(False or True)
print(False or False)
```

```python
not True and False or not False # not이 우선순위
# (not True) and False or (not False) -> False and False or True -> False or True -> True
```

```python
# 논리연산자와 비교연산자 사용 (비교연산자 -> 논리연산자)

print(10 == 10 and 10 != 5)

print(10 > 5 or 10 < 3)

print(not 10 > 5)

print(not 1 is 1.0)
```

```python
(not 10) < 5
# False < 5 -> False < True -> 0 < 1 : False를 0으로 보고 True를 1로 봄
```
